```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

def visualize_stack(stack_data):
    fig, ax = plt.subplots(figsize=(3, 6))
    ax.set_xlim(0, 3); ax.set_ylim(0, len(stack_data)+1)
    ax.axis('off')
    ax.set_title('Stack (LIFO)', fontsize=14, fontweight='bold')
    for i, val in enumerate(stack_data):
        color = '#4ECDC4' if i == len(stack_data)-1 else '#95E1D3'
        rect = patches.FancyBboxPatch((0.5, i+0.1), 2, 0.8,
            boxstyle="round,pad=0.05", facecolor=color, edgecolor='#2C3E50', linewidth=1.5)
        ax.add_patch(rect)
        ax.text(1.5, i+0.5, str(val), ha='center', va='center', fontsize=12, fontweight='bold')
    ax.text(1.5, len(stack_data)+0.5, '← TOP', ha='center', fontsize=10, color='red')
    return fig

def visualize_queue(queue_data):
    fig, ax = plt.subplots(figsize=(len(queue_data)*1.5+1, 3))
    ax.set_xlim(0, len(queue_data)+1); ax.set_ylim(0, 3)
    ax.axis('off')
    ax.set_title('Queue (FIFO)', fontsize=14, fontweight='bold')
    for i, val in enumerate(queue_data):
        color = '#FF6B6B' if i == 0 else '#FFB3B3'
        rect = patches.FancyBboxPatch((i+0.1, 0.5), 0.8, 2,
            boxstyle="round,pad=0.05", facecolor=color, edgecolor='#2C3E50', linewidth=1.5)
        ax.add_patch(rect)
        ax.text(i+0.5, 1.5, str(val), ha='center', va='center', fontsize=11, fontweight='bold')
        if i < len(queue_data)-1:
            ax.annotate('', xy=(i+1.1, 1.5), xytext=(i+0.9, 1.5),
                       arrowprops=dict(arrowstyle='->', color='#2C3E50'))
    ax.text(0.5, 0.2, 'FRONT (dequeue)', ha='center', fontsize=9, color='red')
    ax.text(len(queue_data)-0.5, 0.2, 'REAR (enqueue)', ha='center', fontsize=9, color='blue')
    return fig

stack = [10, 25, 7, 42, 18]
queue = [5, 12, 8, 31, 19, 7]

fig1 = visualize_stack(stack)
fig2 = visualize_queue(queue)

plt.figure(fig1.number)
plt.tight_layout()
plt.savefig('stack_viz.png', dpi=150, bbox_inches='tight')

plt.figure(fig2.number)
plt.tight_layout()
plt.savefig('queue_viz.png', dpi=150, bbox_inches='tight')

plt.show()
print("Stack:", stack)
print("Queue:", queue)
print("\nDataStructures visualized and saved!")
```
